Supervised Learning

Supervised Learning is a type of Machine Learning where the data is labeled. This means the input data already has the correct answers. The model learns from these input-output pairs and tries to predict correct outputs for new data.

Examples:

Spam Detection (Spam or Not Spam)

House Price Prediction

Student Result Prediction

2️⃣ Unsupervised Learning

Unsupervised Learning is used when the data does not have labels. The model tries to find hidden patterns or group similar data points together.

Examples:

Customer Segmentation in marketing

Grouping similar news articles

Market Basket Analysis

3️⃣ Reinforcement Learning

Reinforcement Learning is based on rewards and penalties. An agent learns by interacting with the environment and improving its actions based on feedback.

Examples:

Self-driving cars

Game-playing AI (like chess or video games)

Robot navigation
